## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_19_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# 1. Display number of rows and columns
print("Dataset Shape:", df.shape)

# 2. Display column names
print("Column Names:", df.columns)

# 3. Check data types
df.info()

# 4. Check missing values
print("Missing Values:")
print(df.isnull().sum())

# 5. Count duplicate rows
print("Duplicate Rows:", df.duplicated().sum())
```

4. NummPy practices

Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# Select numerical columns
temperature = df["Temperature_C"].to_numpy()
rainfall = df["Rainfall_MM"].to_numpy()
humidity = df["Humidity_Percent"].to_numpy()

# Calculate average
print("Average Temperature:", np.mean(temperature))
print("Average Rainfall:", np.mean(rainfall))
print("Average Humidity:", np.mean(humidity))

# Calculate maximum
print("Maximum Temperature:", np.max(temperature))
print("Maximum Rainfall:", np.max(rainfall))
print("Maximum Humidity:", np.max(humidity))

# Calculate minimum
print("Minimum Temperature:", np.min(temperature))
print("Minimum Rainfall:", np.min(rainfall))
print("Minimum Humidity:", np.min(humidity))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# 1. Create calculated columns
df["Temperature_F"] = (df["Temperature_C"] * 9/5) + 32

df["Weather_Status"] = df["Rainfall_MM"].apply(
    lambda x: "Rainy" if x > 0 else "No Rain"
)

# 2. Filter records where temperature is above 30°C
hot_days = df[df["Temperature_C"] > 30]

print("Days with Temperature Above 30°C:")
display(hot_days)

# 3. Sort records by temperature
sorted_data = df.sort_values(
    by="Temperature_C", ascending=False
)

print("Top 5 Hottest Days:")
display(sorted_data.head())

# 4. Display complete updated dataset
display(df.head())
```

## 6. Your individual project challenge
**Challenge:** Calculate average, maximum and minimum temperature; identify hottest day and rainy days.

Write your solution below and explain what the result means.


```python
# 1. Calculate average temperature
average_temp = df["Temperature_C"].mean()
print("Average Temperature:", average_temp)

# 2. Calculate maximum temperature
max_temp = df["Temperature_C"].max()
print("Maximum Temperature:", max_temp)

# 3. Calculate minimum temperature
min_temp = df["Temperature_C"].min()
print("Minimum Temperature:", min_temp)

# 4. Identify the hottest day
hottest_day = df.loc[df["Temperature_C"].idxmax()]
print("\nHottest Day:")
display(hottest_day)

# 5. Identify rainy days
rainy_days = df[df["Rainfall_MM"] > 0]
print("\nRainy Days:")
display(rainy_days)

# 6. Count rainy days
print("Total Rainy Days:", len(rainy_days))
```
